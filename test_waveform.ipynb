```python
import warnings 
warnings.filterwarnings("ignore")
import numpy as np 
import matplotlib.pyplot as plt 
import pycbc.waveform as wf 
```

## Anti-glitch frequency-domain waveform 
$$\boxed{
\tilde h(f)
=
A
\exp\left[
-\frac{\mathrm{gbw}}{2}
\left(\ln f-\ln f_0\right)^2
\right]
}$$

```python
flow = 10 
sample_rate = 2048 

f_final = 1024
amp = 1e-21
f0 = 100
gbw = 5

delta_t = 1 / sample_rate 
delta_f = 2 / (sample_rate)

hpf, __loader__ = wf.get_fd_waveform(approximant="antiglitch",
    f_lower=flow,
    delta_f=delta_f,
    f_final=f_final,
    amp=amp,
    f0=f0,
    gbw=gbw)

hpt, _ = wf.get_td_waveform(approximant="antiglitch",
    f_lower=flow,
    delta_t=delta_t,
    f_final=f_final,
    amp=amp,
    f0=f0,
    gbw=gbw)
```

```python
plt.plot(hpf.sample_frequencies, abs(hpf),label='hp')
plt.axvline(x=f0, color='r', linestyle='--', label=f'f0 = {f0} Hz')
plt.grid()
plt.legend()
plt.show()
```

```python
plt.plot(hpt.sample_times, (hpt),label='hp') 
plt.grid()
plt.legend()
plt.show()
```
